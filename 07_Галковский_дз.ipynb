1. Дан список чисел. С помощью map() получить список,
где каждое число из исходного списка переведено в строку.
Пример: на входе – [1, 2, 3], на выходе – [‘1’, ‘2’, ‘3’]

```python
from random import randrange

lst: list[int] = [randrange(100) for _ in range(10)]
print(f"{lst=}")

lst_str: list[str] = list(map(str, lst))
print(f"{lst_str=}")
```

2. Дан список чисел. С помощью filter() получить список
тех элементов из исходного списка, значение которых
больше 0.

```python
from random import randrange

lst: list[int] = [randrange(-100, 100) for _ in range(10)]
print(f"{lst=}")

only_positive_list: list[int] = list(filter(lambda number: number > 0, lst))
print(f"{only_positive_list=}")
```

3. Дан список строк. С помощью filter() получить список
тех строк из исходного списка, которые являются
палиндромами (читаются в обе стороны одинаково, например,
’abcсba’)

```python
lst: list[str] = ["abc", "aba", "aa", "k;dfkfd;l"]
palindromes: list[str] = list(filter(lambda string: string == string[::-1], lst))

print(f"{palindromes=}")
```

4. Сделать декоратор, который измеряет время,
затраченное на выполнение декорируемой функции.

```python
from typing import Callable, Any
from datetime import datetime
from functools import wraps
from time import sleep

def measure_time_function_run(func: Callable) -> Callable:
    @wraps(func)
    def wrapper(*args: Any, **kwargs: Any) -> Any:
        start = datetime.now()
        result = func(*args, **kwargs)
        end = datetime.now()

        total_time = (end - start).total_seconds()
        print(f"{func.__name__=}, works for {total_time}")

        return result

    return wrapper


@measure_time_function_run
def sleeping_function(time_to_sleep: float) -> None:
    sleep(time_to_sleep)


@measure_time_function_run
def iterating_function(iterations: int) -> None:
    for _ in range(iterations):
        pass

sleeping_function(2)
iterating_function(3)
```

5. Используя map() и reduce() посчитать площадь
квартиры, имея на входе характеристики комнат квартиры.
Пример входных данных:
rooms = [ \
    {"name": ”Kitchen", "length": 6, "width": 4}, \
    {"name": ”Room 1", "length": 5.5, "width": 4.5}, \
    {"name": ”Room 2", "length": 5, "width": 4}, \
    {"name": ”Room 3", "length": 7, "width": 6.3}, \
]

```python
from functools import reduce

rooms = [
    {"name": "Kitchen", "length": 6, "width": 4},
    {"name": "Room 1", "length": 5.5, "width": 4.5},
    {"name": "Room 2", "length": 5, "width": 4},
    {"name": "Room 3", "length": 7, "width": 6.3},
]

room_parameters: tuple[float, ...] = tuple(map(
    lambda room_data: room_data.get("length", 0) * room_data.get("width", 0),
    rooms
))
print(f"{rooms=}")
print(f"{room_parameters=}")

area: float = reduce(
    lambda room_parameters1, room_parameters2: room_parameters1 + room_parameters2,
    room_parameters
)
print(f"{area=}")
```
