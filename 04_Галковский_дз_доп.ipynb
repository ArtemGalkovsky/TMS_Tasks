Задача 1:
Напишите программу, которая с помощью модуля random моделирует броски монеты. Программа
принимает на вход количество попыток и выводит результаты бросков: Орел или Решка (каждое
на отдельной строке).

```python
from random import randrange

number_of_throws: int = int(input("Enter number of throws > "))

for index in range(number_of_throws):
    print(index+1, "Heads" if randrange(0, 2) == 0 else "Tails")
```

Задача 4:
Лотерейный билет содержит 7 чисел из диапазона от 1 до 49 (включительно).
Напишите программу, которая с помощью модуля random генерирует 7 различных случайных
чисел для лотерейного билета. Программа должна вывести числа в порядке возрастания на
одной строке через один символ пробела.
Примечание. Убедитесь, что сгенерированные числа не содержат дубликатов

```python
from random import randrange

lottery_ticket: set[int] = set()

while len(lottery_ticket) < 7:
    lottery_ticket.add(randrange(1, 50))

sorted_lottery_ticket: list[int] = sorted(lottery_ticket)
print(*sorted_lottery_ticket)
```

Задача 5:
IP адрес состоит из четырех чисел из диапазона от 0 до 255 (включительно), разделенных точкой.
Напишите функцию generate_ip(), которая с помощью модуля random генерирует и возвращает
случайный корректный IP адрес.

```python
from random import randrange

def generate_ip() -> str:
    return ".".join(str(randrange(0, 256)) for _ in range(4))

print(generate_ip())
```

Задача 7:
Напишите программу, которая с помощью модуля random перемешивает случайным образом
содержимое матрицы (двумерного списка)

```python
from random import shuffle, randrange
from pprint import pprint

matrix: list[list[int, ...], ...] = []

rows_count = randrange(2, 7)
columns_count = randrange(2, 7)

for _ in range(rows_count):
    matrix.append([randrange(0, 10) for _ in range(columns_count)])

print("Initial matrix:")
pprint(matrix)

for row in matrix:
    shuffle(row)

shuffle(matrix)

print("Shuffled matrix:")
pprint(matrix)
```

Задача 10 - Бинго:
Для игры в бинго требуется карточка размером 5 × 5, содержащая различные (уникальные) целые
числа от 1 до 75 (включительно), при этом центральная клетка является пустой (она заполняется
числом 0).
Напишите программу, которая с помощью модуля random генерирует и выводит случайную
карточку для игры в бинго. \
Примечание 1. Для наглядности рекомендуем отводить на вывод каждого числа ровно 3 символа.
Для этого используйте строковый метод ljust().

```python
from random import randrange

field: list[list[int, ...], ...] = []
for _ in range(5):
    field.append([randrange(1, 76) for _ in range(5)])

field[2][2] = 0

print("Field:")
for row in field:
    print(' '.join(str(column).rjust(2) for column in row))
```

Задача 11:
Напишите программу, которая случайным образом назначает каждому ученику его тайного друга,
который будет вместе с ним решать задачи по программированию.
Формат входных данных
На вход программе в первой строке подается число n – общее количество учеников. Далее идут
nnn строк, содержащих имена и фамилии учеников.
Формат выходных данных
Программа должна вывести имя и фамилию ученика (в соответствии с исходным порядком) и имя
и фамилию его тайного друга, разделённые дефисом.

```python
from random import choice

n: int = int(input("Enter number of friends > "))
friends_names: list[str] = [
    input(f"Enter #{friend_index} name >")
    for friend_index in range(n)
]

if len(friends_names) % 2 != 0:
    raise ValueError("U have not even number of friends)))")

while len(friends_names) > 0:
    name = friends_names.pop()

    chosen_friend: str = choice(friends_names)
    friends_names.remove(chosen_friend)

    print(f"{name} - {chosen_friend}")
    print(f"{chosen_friend} - {name}")

```

Задача 12:
Напишите программу, которая с помощью модуля random генерирует n паролей длиной m
символов, состоящих из строчных и прописных английских букв и цифр, кроме тех, которые
легко перепутать между собой: \
«l» (L маленькое); «I» (i большое); «1» (цифра); «o» и «O» (большая и маленькая буквы); «0» (цифра).

Формат входных данных \
На вход программе подаются два числа n и m, каждое на отдельной строке.
Формат выходных данных \
Программа должна вывести n паролей длиной m символов в соответствии с условием задачи,
каждый на отдельной строке.

```python
from random import choices
from string import ascii_letters, digits

PASSWORD_SYMBOLS = "abcdefghijkmnpqrstuvwxyzABCDEFGHJKLMNPQRSTUVWXYZ23456789"

def generate_password(password_length: int, available_symbols: str = ascii_letters + digits) -> str:
    return "".join(choices(available_symbols, k=password_length))

n: int = int(input("Enter number of passwords > "))
m: int = int(input("Enter password length > "))

for i in range(n):
    print(generate_password(m, PASSWORD_SYMBOLS))
```

Задача 13 - генератор паролей из задачи 12 с дополнительным
условием:
В каждом пароле обязательно должна присутствовать хотя бы одна цифра и как минимум по
одной букве в верхнем и нижнем регистре.

```python
from random import choices, choice, shuffle

PASSWORD_DIGITS = "23456789"
PASSWORD_LOWERCASE = "abcdefghijkmnpqrstuvwxyz"
PASSWORD_UPPERCASE = "ABCDEFGHJKLMNPQRSTUVWXYZ"

def generate_password_advanced_security(password_length: int) -> str:
    new_password: list[str] = choices(PASSWORD_LOWERCASE + PASSWORD_UPPERCASE + PASSWORD_DIGITS, k=password_length-3)
    new_password.extend([choice(PASSWORD_DIGITS), choice(PASSWORD_UPPERCASE), choice(PASSWORD_LOWERCASE)])

    shuffle(new_password)

    return "".join(new_password)

n: int = int(input("Enter number of passwords > "))
m: int = int(input("Enter password length > "))

for i in range(n):
    print(generate_password_advanced_security(m))
```
