1. Дан список чисел, отсортированных по возрастанию.
На вход принимается значение, равное одному из элементов
списка. Реализовать функцию, выполняющую рекурсивный
алгоритм бинарного поиска, на выходе программа должна
вывести позицию искомого элемента в исходном списке.

```python
def binary_search(nums: list[int | float], target: int | float, _current_summary_index: int = 0) -> int:
    """
    Search target in list using binary search (recursive).
    :param nums: List of numbers (must be sorted).
    :param target: Number to search for.
    :param _current_summary_index: Used for keeping track of the original index.
    :return: Index of target in the original list, or -1 if not found.
    """
    if not nums:
        return -1

    center_index: int = len(nums) // 2

    if nums[center_index] == target:
        return _current_summary_index + center_index
    elif nums[center_index] > target:
        return binary_search(nums[:center_index], target, _current_summary_index)
    else:
        result = binary_search(nums[center_index + 1:], target, _current_summary_index + center_index + 1)
        return result

lst: list[int] = [2, 40, 42, 43, 53, 57, 58, 64, 69, 88, 89, 98]
print(binary_search(lst, 57))
print(binary_search(lst, 100))
```

2. Программа получает на вход число в десятичной
системе счисления. Реализовать функцию, которая
переводит входное число в двоичную систему счисления.
Допускается реализация функции как в рекурсивном
варианте, так и с итеративным подходом.

```python
def int2binary(positive_integer: int) -> str:
    """
    Convert integer to its binary representation as string.
    :param positive_integer: Positive integer to convert.
    :return: Binary representation as string.
    """

    binary_string = ""
    while positive_integer > 0:
        binary_string = str(positive_integer % 2) + binary_string
        positive_integer //= 2

    return binary_string

for test_number in (1, 2, 100, 200):
    print(f"{int2binary(test_number)}, Real result must be: {bin(test_number)}")
```

3. Программа получает на вход число. Реализовать
функцию, которая определяет, является ли это число простым
(делится только на единицу и на само себя).

```python
from math import sqrt

def is_integer_prime(number: int) -> bool:
    """
    Check if a number is prime.
    :param number: Integer to check.
    :return: True if the number is prime, False otherwise.
    """
    if number < 2:
        return False
    if number == 2:
        return True
    if number % 2 == 0:
        return False

    for divisor in range(3, int(sqrt(number)) + 1, 2):
        if number % divisor == 0:
            return False

    return True

for test_number in (1, 2, 3, 4, 5, 6, 7):
    print(f"{test_number}: {is_integer_prime(test_number)}")
```

4. Программа получает на вход два числа и находит их
НОД (наибольший общий делитель). Пример: на вход
подаются числа 12 и 18, их НОД равен 6.

```python
def find_greatest_common_divisor(number1: int, number2: int) -> int:
    """
    Find the greatest common divisor of two numbers.
    :param number2: int - First number
    :param number1: int - Second number
    :return: int - Greatest common divisor
    """
    max_divisor: int = 1
    for divisor in range(2, min(number1, number2)+1):
        if number1 % divisor == 0 and number2 % divisor == 0:
            max_divisor = divisor

    return max_divisor

for num1, num2 in (
    (1, 2),
    (4, 4),
    (2, 6),
    (3, 8),
    (12, 18),
):
    print(f"{num1} and {num2}: {find_greatest_common_divisor(num1, num2)}")
```

5. Программа получает на вход строку – сообщение и
указание, что нужно сделать: шифровать или дешифровать.
Реализовать две функции: первая шифрует заданное
сообщение шифром Цезаря, вторая расшифровывает. В
зависимости от выбора пользователя (шифровать или
дешифровать) вызывается соответствующая функция,
результат выводится в консоль.

```python
from string import printable

ALPHABET = printable

def caesar_cipher_encode(string: str, shift: int) -> str:
    shift = shift % len(ALPHABET)

    encoded_string = ""
    for letter in string:
        if letter not in ALPHABET:
            raise ValueError(f"The letter '{letter}' in '{string}' is not in the alphabet!")

        new_index = (ALPHABET.index(letter) + shift) % len(ALPHABET)
        encoded_string += ALPHABET[new_index]

    return encoded_string

def caesar_cipher_decode(string: str, shift: int) -> str:
    shift = shift % len(ALPHABET)

    decoded_string = ""
    for letter in string:
        if letter not in ALPHABET:
            raise ValueError(f"The letter '{letter}' in '{string}' is not in the alphabet!")

        new_index = (ALPHABET.index(letter) - shift) % len(ALPHABET)
        decoded_string += ALPHABET[new_index]

    return decoded_string


print(caesar_cipher_encode("HELLO123", 10))
print(caesar_cipher_decode(caesar_cipher_encode("HELLO123", 10), 10))

try:
    print(caesar_cipher_encode("123русский да", 10))
except ValueError as e:
    print(f"Error: {e}")
```
