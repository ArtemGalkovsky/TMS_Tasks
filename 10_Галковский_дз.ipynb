1. Создайте класс Soda (газировка). Для инициализации
есть параметр, который определяет вкус газировки. При
инициализации этот параметр можно задавать, а можно и не
задавать. Реализовать метод строковой репрезентации,
который возвращает строку вроде «У вас газировка с
<клубничным> вкусом», если вкус задан. Если вкус не задан,
метод должен возвращать строку «У вас обычная газировка».

```python
class Soda:
    def __init__(self, taste: str | None = None) -> None:
        self.taste: str = taste

    def __str__(self):
        if self.taste is None:
            return "У вас обычная газировка"

        return f"У вас газировка с {self.taste} вкусом"

soda1 = Soda()
print(soda1)

soda2 = Soda("Клубничка")
print(soda2)
```

2. Напишите программу с классом Math. При
инициализации атрибутов нет. Реализовать методы addition,
subtraction, multiplication и division. При передаче в методы
двух числовых параметров нужно производить с
параметрами соответствующие действия и печатать ответ.

```python
# 3.12+ syntax
type Number = int | float

class Math:
    @staticmethod
    def add(a: Number, b: Number) -> Number:
        return a + b

    @staticmethod
    def subtract(a: Number, b: Number) -> Number:
        return a - b

    @staticmethod
    def multiply(a: Number, b: Number) -> Number:
        return a * b

    @staticmethod
    def divide(a: Number, b: Number) -> Number:
        return a / b

print(Math.add(1, 237.2))
print(Math.subtract(1, 237.2))
print(Math.multiply(1, 237.2))
print(Math.divide(1, 237.2))
```

3. Программа с классом Car. При инициализации объекта
ему должны задаваться атрибуты color, type и year.
Реализовать пять методов. Запуск автомобиля – выводит
строку «Автомобиль заведён». Отключение автомобиля –
выводит строку «Автомобиль заглушен». Методы для
присвоения автомобилю года выпуска, типа и цвета.

```python
class Car:
    def __init__(self, color: str, type_: str, year: int) -> None:
        self._color: str = color
        self._type: str = type_
        self._year: int = year

    def start_engine(self) -> None:
        print(f"Автомобиль {self} заведён")

    def stop_engine(self) -> None:
        print(f"Автомобиль {self} заглушен")

    def __str__(self) -> str:
        return f"Автомобиль цвета {self._color}, типа {self._type} и {self._year} года выпуска"

    @property
    def color(self) -> str:
        return self._color

    @color.setter
    def color(self, new_color: str) -> None:
        self._color = new_color

    @property
    def type_(self) -> str:
        return self._type

    @type_.setter
    def type_(self, new_type: str) -> None:
        self._type = new_type

    @property
    def year(self) -> int:
        return self._year

    @year.setter
    def year(self, new_year: int) -> None:
        self._year = new_year

auto = Car("Yellow", "BMW", 2021)
auto.start_engine()
auto.stop_engine()

auto.color = "Red"
auto.type_ = "BMG"
auto.year = 2020

print("Car:", auto)
print("Color:", auto.color)
print("Type:", auto.type_)
print("Year:", auto.year)
```

4. Программа с классом Sphere для представления сферы
в трёхмерном пространстве. Реализовать методы: \
● конструктор, принимающий 4 числа: радиус и координаты
центра сферы x, y, z. Если конструктор вызывается без
аргументов, создать объект сферы с единичным радиусом
и центром в начале координат. Если конструктор
вызывается только с радиусом, создать объект с
соответствующим радиусом и центром в начале
координат \
● метод get_volume(), возвращающий число – объем шара,
ограниченного текущей сферой \
● метод get_square(), возвращающий число – площадь
внешней поверхности сферы \
● метод get_radius(), возвращающий число – радиус текущей
сферы \
● метод get_center(), возвращающий кортеж с координатами
центра сферы \
● метод set_radius(radius), который принимает новое
значение радиуса, меняет радиус текущей сферы и ничего
не возвращает \
● метод set_center(x, y, z), который принимает новые
значения для координат центра радиуса, меняет
координаты текущей сферы и ничего не возвращает \
● метод is_point_inside(x, y, z), который принимает
координаты некой точки в трёхмерном пространстве и
возвращает True или False в зависимости от того,
находится ли точка внутри сферы

```python
from math import pi

# 3.12+ syntax
type Number = int | float

class Sphere:
    def __init__(self, radius: Number = 1, x: Number = 0, y: Number = 0, z: Number = 0) -> None:
        self._radius: Number = radius
        self._x: Number = x
        self._y: Number = y
        self._z: Number = z

    def get_volume(self) -> Number:
        return 4/3 * pi * self._radius ** 3

    def get_square(self) -> Number:
        return 4 * pi * self._radius ** 2

    def get_radius(self):
        return self._radius

    def get_center(self):
        return self._x, self._y, self._z

    def set_radius(self, new_radius: Number) -> None:
        if new_radius < 0:
            raise ValueError("Radius cannot be negative!")
        elif not isinstance(new_radius, Number.__value__):
            raise TypeError("Radius must be a number!")

        self._radius = new_radius

    def set_center(self, x: Number, y: Number, z: Number) -> None:
        if (
                not isinstance(x, Number.__value__)
                or not isinstance(y, Number.__value__)
                or not isinstance(z, Number.__value__)
        ):
            raise TypeError("x, y, z must be numbers!")

        self._x = x
        self._y = y
        self._z = z

    def is_point_inside(self, x: Number, y: Number, z: Number) -> bool:
        distance_squared = (x - self._x) ** 2 + (y - self._y) ** 2 + (z - self._z) ** 2
        return distance_squared <= self._radius ** 2

    def __str__(self) -> str:
        return f"Сфера радиусом {self._radius} и координатами ({self._x}, {self._y}, {self._z})."

sphere_none = Sphere()
print(sphere_none)

sphere_only_radius = Sphere(10)
print(sphere_only_radius)

sphere_full_args = Sphere(10, 2, 3, 4)
print(sphere_full_args)

sphere_full_args.set_radius(5)
print("set_radius(5)", sphere_full_args)

sphere_full_args.set_center(5, 3, 2)
print("set_center(5, 3, 2)", sphere_full_args)

print(f"{sphere_full_args.is_point_inside(5, 3, 2) = }")
print(f"{sphere_full_args.is_point_inside(10, 10, 10) = }")
```

5. Разработать класс SuperStr, который наследует
функциональность стандартного типа str и содержит два
новых метода: \
● метод is_repeatance(s), который принимает некоторую
строку и возвращает True или False в зависимости от того,
может ли текущая строка быть получена целым
количеством повторов строки s. Считать, что пустая
строка не содержит повторов \
● метод is_palindrom(), который возвращает True или False в
зависимости от того, является ли строка палиндромом вне
зависимости от регистра. Пустую строку считать
палиндромом.

```python
class SuperStr(str):
    def __new__(cls, *args, **kwargs):
        return str.__new__(cls, *args, **kwargs)

    def is_repeatance(self, s: str) -> bool:
        if not s or len(self) % len(s) != 0:
            return False

        return self == s * (len(self) // len(s))

    def is_palindrome(self):
        if self[::-1].lower() == self.lower():
            return True

        return False

super_str = SuperStr("hello")

print(f"{super_str.is_repeatance('h') = }")
print(f"{super_str.is_palindrome() = }")

super_str_palindrome_and_repeatance = SuperStr("111")
print(f"{super_str_palindrome_and_repeatance.is_repeatance('1') = }")
print(f"{super_str_palindrome_and_repeatance.is_palindrome() = }")
```
