1. Учимся работать с формулами.
Используя модуль math, вычислите значения по формулам.
Значения a, b, x ввести с клавиатуры, вывести 4
рассчитанных по формулам значения. Сделать красиво
оформленные ввод и вывод данных.

```python
from math import sin, cos, sqrt, cbrt

a, b, x = map(float, input("Enter a, b, x separated by space: ").split())

power2_a = a*a

y_a = (power2_a / 3
       +
       (power2_a + 4) / b
       +
       sqrt(power2_a + 4) / 4
       +
       sqrt((power2_a + 4) ** 3) / 4)

y_b = cos(x) + sin(x)

y_c = cbrt(
    cos(x*x) ** 2
    +
    sin(2*x - 1) ** 2
)

y_d = 5*x + 3 * x*x * sqrt(1 + sin(x)**2)

print(f"{a = }, {b = }, {x = }.")
print(f"{y_a = }")
print(f"{y_b = }")
print(f"{y_c = }")
print(f"{y_d = }")
```

2. «Играл с кредитами – проиграл»
m – ежемесячная выплата
i – годовая процентная ставка
p – месячная процентная ставка
s – сумма займа
n – количество месяцев, на
которые взят кредит. \
Пользователь вводит с клавиатуры i, s и n, программа
должна рассчитать размер ежемесячной выплаты по
формуле, а также найти, сколько пользователь всего
заплатит банку и сколько составит переплата. Сделать
ввод/вывод данных красивым интерактивом

```python
i = float(input("Введите годовую процентную ставку i в %: "))
s = float(input("Введите сумму займа s: "))
n = int(input("Введите количество месяцев n: "))

p = i / 100 / 12
m = s * p * (1 + p)**n / ((1 + p)**n - 1)

total = m * n
overpayment = total - s

print(f"Ежемесячная выплата: {m}")
print(f"Общая сумма выплат: {total}")
print(f"Переплата: {overpayment}")
```

3. «Интерстеллар»
L - длина года на планете
R – радиус орбиты планеты (млн.
км.)
𝒗 – орбитальная скорость (км/ч)
Пользователь вводит с клавиатуры информацию о двух
планетах: радиус и орбитальную скорость. Программа
должна посчитать и вывести длину года на первой и второй
планетах, а также вывести, правда ли, что на первой планете
год длиннее, чем на второй. Сделать ввод/вывод данных
красивым интерактивом

```python
from math import pi

print("Введите данные для первой планеты > ")
R1 = float(input("Радиус орбиты > "))
v1 = float(input("Орбитальная скорость > "))

print("\nВведите данные для второй планеты > ")
R2 = float(input("Радиус орбиты > "))
v2 = float(input("Орбитальная скорость > "))

L1 = (2 * R2 * pi) / v1
L2 = (2 * R1 * pi) / v2

print(f"\nДлина года на первой планете: {L1} часов")
print(f"Длина года на второй планете: {L2} часов")

if L1 > L2:
    print("На первой планете год длиннее!")
elif L1 < L2:
    print("На второй планете год длиннее!")
else:
    print("Годы на обеих планетах одинаковые!")
```
